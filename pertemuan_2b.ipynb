```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
import random

# Membuat black image sebagai kanvas
black_img = np.zeros(shape=(512, 512, 3), dtype=np.uint8)

# Membuat 5 kotak dengan ukuran, posisi, dan warna acak
for i in range(5):
    # titik awal (pt1) acak
    x1 = random.randint(0, 400)
    y1 = random.randint(0, 400)

    # ukuran kotak acak (lebar & tinggi antara 30-150 px)
    w = random.randint(30, 150)
    h = random.randint(30, 150)
    x2 = min(x1 + w, 512)
    y2 = min(y1 + h, 512)

    # warna acak (R, G, B)
    color = (random.randint(0, 255), random.randint(0, 255), random.randint(0, 255))

    cv.rectangle(black_img, pt1=(x1, y1), pt2=(x2, y2), color=color, thickness=-1)  # -1 = terisi penuh

plt.imshow(black_img)
plt.title("5 kotak dengan ukuran dan warna acak")
plt.show()
```

```python
import cv2 as cv
import matplotlib.pyplot as plt

# Baca image dan konversi ke RGB
img = cv.imread("/content/drive/MyDrive/PCVK/KTM lama.png")
imgRGB = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# Flip horizontal (kiri-kanan terbalik)
imgFlipHorizontal = cv.flip(imgRGB, 1)

# Flip vertical (atas-bawah terbalik)
imgFlipVertical = cv.flip(imgRGB, 0)

# Flip kedua arah sekaligus (180 derajat)
imgFlipBoth = cv.flip(imgRGB, -1)

# Tampilkan ketiganya berdampingan
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
axs[0].imshow(imgFlipHorizontal)
axs[0].set_title("Flip Horizontal")
axs[1].imshow(imgFlipVertical)
axs[1].set_title("Flip Vertical")
axs[2].imshow(imgFlipBoth)
axs[2].set_title("Flip Horizontal + Vertical")
plt.show()
```
