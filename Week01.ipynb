```python


import cv2
import numpy as np


def beri_label_identitas(citra, teks,
                          warna_bar=(0, 0, 0),
                          warna_teks=(255, 255, 255)):

    tinggi, lebar = citra.shape[:2]
    tinggi_bar = tinggi // 12
    kanvas = np.zeros((tinggi + tinggi_bar, lebar, citra.shape[2]),
                       dtype=citra.dtype)
    kanvas[0:tinggi, 0:lebar] = citra
    kanvas[tinggi:tinggi + tinggi_bar, 0:lebar] = warna_bar
    skala_font = tinggi_bar / 50.0
    tebal_font = max(1, round(tinggi_bar / 25))
    margin_x = int(lebar * 0.02)
    lebar_maks_teks = lebar - 2 * margin_x
    (lebar_teks, _), _ = cv2.getTextSize(
        teks, cv2.FONT_HERSHEY_SIMPLEX, skala_font, tebal_font)
    if lebar_teks > lebar_maks_teks:
        skala_font *= lebar_maks_teks / lebar_teks
        tebal_font = max(1, round(tebal_font * lebar_maks_teks / lebar_teks))
    (_, tinggi_teks), _ = cv2.getTextSize(
        teks, cv2.FONT_HERSHEY_SIMPLEX, skala_font, tebal_font)
    x_teks = margin_x
    y_teks = tinggi + tinggi_bar // 2 + tinggi_teks // 2

    cv2.putText(kanvas, teks, (x_teks, y_teks),
                cv2.FONT_HERSHEY_SIMPLEX, skala_font,
                warna_teks, tebal_font, cv2.LINE_AA)

    print(f"Ukuran masukan : {lebar} x {tinggi} px")
    print(f"Tinggi bar     : {tinggi} // 12 = {tinggi_bar} px")
    print(f"Skala font     : {skala_font:.2f}  | Ketebalan: {tebal_font}")

    return kanvas


if __name__ == "__main__":
    path_foto_1 = "ktp.png"
    path_foto_2 = "KTM CONTOH.png"
    for path in (path_foto_1, path_foto_2):
        citra = cv2.imread(path)
        if citra is None:
            print(f"[!] Tidak dapat membaca: {path}")
            continue

        hasil = beri_label_identitas(citra, "Dokumentasi PCVK - 2026")

        nama_keluaran = "label_" + path
        cv2.imwrite(nama_keluaran, hasil)
        print(f"Disimpan sebagai: {nama_keluaran}\n")
```
